```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
titanic = pd.read_csv("train.csv")
```

```python
titanic.head()
```

## First Hypothesis

Determine if the survival rate is associated to the class of passenger

```python
surv_class = pd.crosstab(titanic['Pclass'], titanic['Survived'])
surv_class = surv_class.rename(columns = {0: 'No', 1:'Yes'})
surv_class = surv_class[["Yes", "No"]]
surv_class
```

```python
surv_class.plot(kind = "bar", rot =0)
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Survival by Passenger Class")
plt.savefig("Survival by Passenger Class")
```

```python
surv_rate = titanic.groupby('Pclass')['Survived'].mean() *100
surv_rate
```

```python
surv_rate.plot(kind = "bar", rot = 0, legend = False, color = 'green')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate in %')
plt.title("Survival Rate by Class")
plt.savefig("Survival Rate by Class")
```

## Second Hypothesis
Determine if the survival rate is associated to the gender

```python
surv_gender = pd.crosstab(titanic['Sex'], titanic['Survived'])
surv_gender = surv_gender.rename(columns = {0: 'No', 1:'Yes'})
surv_gender = surv_gender[["Yes", "No"]]
surv_gender
```

```python
surv_gender.plot(kind = "bar", rot =0, legend = False)
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.title("Survival by Gender")
plt.legend()
plt.savefig("Survival by Gender")
```

```python
surv_rate_gender = titanic.groupby('Sex')['Survived'].mean() * 100
surv_rate_gender
```

```python
surv_rate_gender.plot(kind = "bar", rot = 0, legend = False, color = 'pink')
plt.xlabel('Gender')
plt.ylabel('Survival Rate in %')
plt.title("Survival Rate by Gender")
plt.savefig("Survival Rate by Gender")
```

## Third Hypothesis
Determine if the survival rate is associated to age


```python
titanic["AgeGroup"] = pd.cut(titanic["Age"], bins=range(0, 81, 10))
surv_rate_age = titanic.groupby("AgeGroup")["Survived"].mean() * 100
```

```python
surv_rate_age.plot(kind = "bar", color = "skyblue", rot = 0, legend = False)

plt.xlabel("Age Group")
plt.ylabel("Survival Rate in %")
plt.ylim(0,100)
plt.title("Survival Rate by Age")
plt.savefig("Survival Rate by Age")
```
